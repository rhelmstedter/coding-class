## Two Tools You Might Want

**A story is longer than one line.** A triple-quoted f-string lets you write it the way it should read, line breaks and all:

```python
print(f"""{name} opened the door and found {plural_noun} everywhere.
Nobody in {place} could explain it.""")
```

Three quote marks to open, three to close. Everything between them prints exactly as you typed it. Several separate `print()` lines work just as well if you prefer.

**Reusing a word costs nothing.** Once a word is in a variable you can drop it into the story as many times as you like:

```python
print(f"{name} tried. {name} failed. {name} tried again.")
```

You asked once. You can use it as often as you want. Not required, but it makes a story read better.
